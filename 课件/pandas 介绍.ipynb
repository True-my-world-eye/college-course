### Series 对象创建的几种方式

Series 对象的创建基本上符合下面的语法：

```python
>>> pd.Series(data, index=index)
```

data 可以是numpy array，list，index默认的值为int 型的索引

````python
pd.Series([2, 4, 6])
````

````python
pd.Series(5, index=[100, 200, 300])
````

````python
pd.Series({2:'a', 1:'b', 3:'c'})
````

````python
pd.Series({2:'a', 1:'b', 3:'c'}, index=[3, 2])
````

## 数据的索引和选取

### Series 对象

````python
data = pd.Series([0.25, 0.5, 0.75, 1.0],
                 index=['a', 'b', 'c', 'd'])
data
````

````python
data['b']
````

````python
'a' in data
````

````python
data.keys()
````

````python
list(data.items())
````

````python
# slicing by explicit index
data['a':'c']
````

````python
# slicing by implicit integer index
data[0:2]
````

````python
data > 0.3
````

````python
# masking
data[(data > 0.3) & (data < 0.8)]
````

````python
indx = ['a', 'c']
````

````python
# fancy indexing
data[indx]
````

### Dataframe 对象

````python
area = pd.Series({'California': 423967, 'Texas': 695662,
                  'New York': 141297, 'Florida': 170312,
                  'Illinois': 149995})
pop = pd.Series({'California': 38332521, 'Texas': 26448193,
                 'New York': 19651127, 'Florida': 19552860,
                 'Illinois': 12882135})
data = pd.DataFrame({'area':area, 'pop':pop})
data
````

````python
data['area']
````

````python
data.area
````

````python
data.area is data['area']
````

````python
data['density'] = data['pop'] / data['area']
data
````

````python
data.iloc[0]
````

````python
data.values[0]
````

````python
data.T
````

````python
data.values[0]
````

````python
# equivalent to the following
data.iloc[0]
````

````python
data.iloc[:3, :2]
````

````python
data.loc[:'Illinois', :'pop']
````

````python
data.loc[data.density > 100, ['pop', 'density']]
````

````python
data.iloc[0, 2] = 90
data
````
